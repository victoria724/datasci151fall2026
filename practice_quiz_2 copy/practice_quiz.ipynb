**Question (B)** Defining a variable through a piecewise-defined expression

<font size = "3">

Consider the piecewise-defined expression

$$
y = 
\begin{cases}
0 & \text{if } x \le 0 \\[6pt]
\dfrac{1}{2}x & \text{if } 0 < x \le \tfrac{1}{2} \\[10pt]
\dfrac{1}{2}(1 - x) & \text{if } \tfrac{1}{2} < x \le 1 \\[6pt]
0 & \text{if } 1 < x
\end{cases}
$$

In the code cell below, a random number `x` is generated uniformly
between $-1/2$ and $3/2$. Using `if`, `elif`, and/or `else` statements, write code that defines a variable `y` that is determined by the expression above. Then print it to the screen.

````python
import random 

x = random.uniform(-0.5, 1.5)

# your code below
if x <= 0:
    y = 0
elif x <= 1/2:
    y = 1/2 * x
elif x <=1:
    y = 1/2*(1-x)
else:
    y = 0


print(y)
````

**Question (C)** IF/ELIF/ELSE statements

<font size = "3">

- In the beginning of the code cell below, two strings are created, where the first word is one of three random choices: 
```python
    title_py = "Introduction/Intro/Guide to Python"
    title_r = "Introduction/Intro/Guide to R"
```

- Assign the string "Introduction" to a variable named `keyword`
- Create an if/elif/else block that tests whether `keyword` appears in the two strings.  

- Depending on the outcome (appears in both strings, appears in `title_py` only, appears in `title_r` only, doesn't appear in either string) print an appropriate message to the screen.

HINT: If the creation of `title_py` and `title_r` is confusing, add print statements to view the two strings and run the cell 3-4 times to see some of the possible situations

````python
### Code to create strings. DO NOT CHANGE!!
import random 
first_word_choices = ["Introduction", "Intro", "Guide"]
selected_words = random.choices(first_word_choices, k = 2)

title_py = selected_words[0] + " to Python"
title_r = selected_words[1] + " to R"


### Your solution below
keyword = "Introduction"
if keyword in title_py:
    if keyword in title_r:
        print("In both")
    else:
        print("In title_py")
elif keyword in title_r:
    print("In title_r")
else:
    print("No")

print(title_py)
print(title_r)
````

**Question (D)** Looping over a list

<font size = "3">

In the cell below, 3 scatter plots are generated using the car data from "files/features_clean.csv". The scatter plots show:

- displacement vs. mpg
- weight vs. mpg
- acceleration vs. mpg

````python
import pandas as pd 
import matplotlib.pyplot as plt 

# Setup code
df_cars = pd.read_csv("files/features_clean.csv")
y_var = 'mpg'

# unrolled loop

x_var = 'displacement'
plt.scatter(df_cars[x_var], df_cars[y_var])
plt.xlabel(x_var)
plt.ylabel(y_var)
plt.title(x_var + " vs. " + y_var)
plt.show()

x_var = 'weight'
plt.scatter(df_cars[x_var], df_cars[y_var])
plt.xlabel(x_var)
plt.ylabel(y_var)
plt.title(x_var + " vs. " + y_var)
plt.show()


x_var = 'acceleration'
plt.scatter(df_cars[x_var], df_cars[y_var])
plt.xlabel(x_var)
plt.ylabel(y_var)
plt.title(x_var + " vs. " + y_var)
plt.show()
````

<font size = "3">

The code in the previous cell was written as an "unrolled" loop. Complete the code cell below so that the same 3 plots are generated inside a "for" loop.



````python
# Setup code
df_cars = pd.read_csv("files/features_clean.csv")
y_var = 'mpg'
plot_variables = ['displacement', 'weight', 'acceleration']

# Your code below

for i in plot_variables:
    plt.scatter(df_cars[i],df_cars[y_var])
    plt.xlabel(i)
    plt.title(i + " v.s. " + y_var)
    plt.show()
````
